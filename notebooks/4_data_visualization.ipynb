# Data Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/processed/3_engineered_data.csv', sep=';', encoding='utf-8')
df.head()
```

```python
filter_df = df[df['SCORE'] == 0]
```

### SURFACE

```python
plt.bar(data=df, x=df['SURFACE'].value_counts().index, height=df['SURFACE'].value_counts().values)
```

### Tourney Level

```python
plt.bar(data=df, x=df['TOURNEY_LEVEL'].value_counts().index, height=df['TOURNEY_LEVEL'].value_counts().values)
```

### Player Hand

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].bar(data=filter_df, x=df['PLAYER_0_HAND'].value_counts().index, height=df['PLAYER_0_HAND'].value_counts().values)
ax[0].set_title('Winner Hand')

ax[1].bar(data=filter_df, x=df['PLAYER_1_HAND'].value_counts().index, height=df['PLAYER_1_HAND'].value_counts().values)
ax[1].set_title('Loser Hand')
```

- As evidantiated by these graphs, there is no correlation between player hand and score.
- Hand of winner and loser follows the same pattern and distribution and values

### Player Height

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
ax[0].boxplot(x=filter_df['PLAYER_0_HT'])
ax[0].set_title('Winner Height')
ax[1].boxplot(x=filter_df['PLAYER_1_HT'])
ax[1].set_title('Loser Height')
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].hist(filter_df['PLAYER_0_HT'], bins=15)
ax[0].set_title('Winner Height')
ax[1].hist(filter_df['PLAYER_1_HT'], bins=15, color='orange')
ax[1].set_title('Loser Height')
plt.show()
```

```python
#Correlation between PLAYER_0_HT and PLAYER_1_HT
plt.figure()
sns.scatterplot(x=df['PLAYER_0_HT'], y=df['PLAYER_1_HT'], hue=df['SCORE'])
plt.show()
```

```python
#Correlation between PLAYER_HT and PLAYER_RATING
plt.figure()
sns.scatterplot(x=df['PLAYER_0_HT'], y=df['PLAYER_0_RATING'], hue=df['SCORE'])
plt.show()
```

```python
#Correlation between HT_DIFF and SCORE
plt.figure()
sns.violinplot(data=df, x=df['HT_DIFF'], hue=df['SCORE'])
plt.show()
```

- As evidantiated by these graphs, there is no correlation between player height and score.
- Height of winner and loser follows the same pattern and distribution and values

```python
# pair plot with numerical numbers
# Filtering by features that splits score very well
sample_df = df[['SCORE', 'HT_DIFF', 'AGE_DIFF',
       'RATING_DIFF', 'SURFACE_RATING_DIFF', 'WIN_RATE_DIFF', 
       'LAST_10_WIN_DIFF', 'LAST_25_WIN_DIFF', 'LAST_50_WIN_DIFF',
       'LAST_100_WIN_DIFF', 'WIN_COMBINED_DIFF']]
sample_df = sample_df.sample(n=5000, random_state=42)
sample_df
```

```python
sns.pairplot(sample_df, hue='SCORE')
plt.show()
```

```python
sns.clustermap(sample_df.corr(), cmap='coolwarm', center=0, figsize=(12, 12))
plt.show()
```

# Result of Data Visualization

- Some features have no correlation with the, like PLAYER_HAND, PLAYER_AGE and PLAYER_HT

- Features like AGE_DIFF, HT_DIFF correlate with others features that correlate with all others features

- RATING_DIFF, SURFACE_RATING_DIFF, LAST_50_WIN_DIFF and LAST_100_WIN_DIFF splits the score result very well

- LAST_10 and LAST_25 splits well too but not as good as LAST_50 and LAST_100
