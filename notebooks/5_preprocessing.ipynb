# Data Preprocessing

- Feature Selection
- Data split

    - Feature
    - Target

- Standardize features
- Dimensionality Reduction


```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, RobustScaler
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
df = pd.read_csv('../data/processed/3_engineered_data.csv', sep=';', encoding='utf-8')
df
```

## Feature Selection

- Data analysis done in **data_visualization.ipynb**

```python
df.columns
```

```python
df = df.drop(df[['SURFACE', 'TOURNEY_LEVEL', 'PLAYER_0_ID',
       'PLAYER_0_HAND', 'PLAYER_0_HT', 'PLAYER_0_AGE', 'PLAYER_1_ID',
       'PLAYER_1_HAND', 'PLAYER_1_HT', 'PLAYER_1_AGE', 'ROUND']], axis=1)
```

```python
df.describe()
```

## Data Split

```python
feature = df.drop('SCORE', axis=1)
feature.head(3)
```

```python
target = df[['SCORE']]
target.value_counts()
```

```python
feature.to_csv('../data/processed/5.1_data_feature.csv', sep=';', encoding='utf-8', index=False)
target.to_csv('../data/processed/5.1_data_target.csv', sep=';', encoding='utf-8', index=False)
```

## Standardize Features

#### StandardScaler (mean = 0, std = 1)

```python
scaler = StandardScaler()
```

```python
feature_std_esc = scaler.fit_transform(feature.drop(['TOURNEY_DATE'], axis=1))
feature_std_esc = pd.DataFrame(feature_std_esc)
```

```python
feature_std_esc
```

```python
feature_std_esc['TOURNEY_DATE'] = feature['TOURNEY_DATE']
feature_std_esc.head(3)
```

```python
feature_std_esc.describe()
```

#### RobustScaler (better with outliers)

```python
robust = RobustScaler()
```

```python
feature_robust_esc = robust.fit_transform(feature.drop(['TOURNEY_DATE'], axis=1))
feature_robust_esc = pd.DataFrame(feature_robust_esc)
```

```python
feature_robust_esc['TOURNEY_DATE'] = feature['TOURNEY_DATE']
feature_robust_esc.head(3)
```

```python
feature_robust_esc.describe()
```

```python
feature_std_esc.to_csv('../data/processed/5.2_data_feature_std_esc.csv', sep=';', encoding='utf-8', index=False)
feature_robust_esc.to_csv('../data/processed/5.2_data_feature_robust_esc.csv', sep=';', encoding='utf-8', index=False)
```

## Dimensionality Reduction (Using RobustScaler Dataset)

#### PCA

```python
pca = PCA(n_components=15, svd_solver='arpack', random_state=0)
```

```python
pca_robust = pca.fit_transform(feature_robust_esc.drop(['TOURNEY_DATE'], axis=1))
```

```python
pca_robust.shape
```

```python
pca.explained_variance_ratio_.sum()
```

```python
pca_robust = pd.DataFrame(pca_robust)
pca_robust['TOURNEY_DATE'] = feature['TOURNEY_DATE']
pca_robust
```

#### LDA

```python
lda = LinearDiscriminantAnalysis(n_components=1)
```

```python
lda_robust = lda.fit_transform(X=feature_robust_esc.drop(['TOURNEY_DATE'], axis=1), y=target['SCORE'])
lda_robust
```

```python
lda_robust = pd.DataFrame(lda_robust)
lda_robust['TOURNEY_DATE'] = feature['TOURNEY_DATE']
lda_robust
```

```python
pca_robust.to_csv('../data/processed/5.3_data_feature_robust_pca.csv', sep=';', encoding='utf-8', index=False)
lda_robust.to_csv('../data/processed/5.3_data_feature_robust_lda.csv', sep=';', encoding='utf-8', index=False)
```
